```python
import pandas as opcoesPanda

notasAluno_DataFrame = opcoesPanda.DataFrame({
    "Nome": ["João", "Maria", "Pedro", "Ana"],
    "Nota1": [7.5, 8.0, 6.0, 9.0],   
    "Nota2": [8.0, 7.5, 8.5, 8.0],
    "Nota3": [9.0, 8.5, 7.0, 8.5],
    "Nota4": [7.0, 8.0, 6.5, 9.0]
})

display(notasAluno_DataFrame)
```

```python
# Adicionando colunas
notasAluno_DataFrame["Média"] = (notasAluno_DataFrame[["Nota1", "Nota2", "Nota3", "Nota4"]].mean(axis=1))

#Criando valor padrão
notasAluno_DataFrame["Faltas"] = 5

#Localizando e alterando a note do Pedro
notasAluno_DataFrame.loc[2, "Nota2"] = 50

display(notasAluno_DataFrame)
```
